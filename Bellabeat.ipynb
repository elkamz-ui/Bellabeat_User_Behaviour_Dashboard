```python
import pandas as pd
import random as rd
df_daily_activity = pd.read_csv("Daily.Activity.csv")
df_sleep_day = pd.read_csv("Sleep.Day.csv")
df_weight_log = pd.read_csv("Weight.Log.csv")
```

```python
df_weight_log
```

```python
df_weight_log = df_weight_log.dropna(subset="Customer_ID")
df_weight_log["Date"] = pd.to_datetime(df_weight_log["Date"])
df_weight_log["Day_of_Week"] = df_weight_log["Date"].dt.day_name()
df_weight_log["Date.1"] = df_weight_log["Date"].dt.date
df_weight_log["Time"] = df_weight_log["Date"].dt.time
df_weight_log["Weight_Kg"] = pd.to_numeric(df_weight_log["Weight_Kg"])
df_weight_log["Weight_Pounds"] = pd.to_numeric(df_weight_log["Weight_Pounds"])
df_weight_log["Fat"] = pd.to_numeric(df_weight_log["Fat"])
df_weight_log["BMI"] = pd.to_numeric(df_weight_log["BMI"])
df_weight_log["Fat"] = df_weight_log["Fat"].apply(lambda x:rd.randint(10,40) if pd.isna(x) else x)
df_weight_log.info()
```

```python
df_daily_activity["Activity_Date"] = pd.to_datetime(df_daily_activity["Activity_Date"])
df_daily_activity["Day_of_Week"] = df_daily_activity["Activity_Date"].dt.day_name()
df_daily_activity
```

```python
df_sleep_day["Sleep_Day"] = pd.to_datetime(df_sleep_day["Sleep_Day"])
df_sleep_day["Day_of_Week"] = df_sleep_day["Sleep_Day"].dt.day_name()
df_sleep_day
```
